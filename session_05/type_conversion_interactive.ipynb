## 1️⃣ The Problem — why types matter

```python
user_input = '500'
print(f'Type: {type(user_input)}')

try:
    result = user_input + 100
except TypeError as e:
    print(f'ERROR: {e}')
    print('Solution: convert first!')
```

## 2️⃣ str → int — text to whole number

```python
credit_score = int('742')
otp = int('4829')

print(f'Credit score : {credit_score}  {type(credit_score)}')
print(f'Score + 50   : {credit_score + 50}')
print(f'OTP match    : {otp == 4829}')
```

## 3️⃣ int → float — whole number to decimal

```python
per_person = float(600)
tip   = per_person * 0.10
total = per_person + tip

print(f'Per person : ₹{per_person}')
print(f'Tip (10%)  : ₹{tip}')
print(f'Total      : ₹{total}')
```

## 4️⃣ int/float → str — number to text

```python
amount = 240
restaurant = 'Biryani Blues'
mins = 32

sms = f'Your order from {restaurant} of ₹{amount} arrives in {mins} mins.'
print(f'📱 SMS: {sms}')
```

## 5️⃣ = vs == — the most confusing difference

```python
score = 742

print(f'score == 742 : {score == 742}')
print(f'score == 800 : {score == 800}')
print('= stores a value. == checks a value.')
```

## 6️⃣ Bonus — ** and // operators

```python
print(f'2 ** 3 = {2**3}  # to the power of')
print(f'7 // 2 = {7//2}  # floor division')
print(f'7 %  2 = {7%2}  # leftover')
```

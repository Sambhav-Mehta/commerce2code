## Control Flow — if, elif, else

> Swiggy charges ₹0 sometimes. ₹30 other times. ₹60 when it rains. That's not random. That's control flow.

## 1️⃣ The Simplest Decision — if/else
One condition. Two outcomes.

```python
# Does the user have Swiggy One?
has_swiggy_one = False

if has_swiggy_one:
    print('✅ Free delivery — Swiggy One member')
else:
    print('❌ Delivery fee applies')

# Try changing False to True and run again
```

## 2️⃣ Multiple Conditions — if/elif/else

Real apps check many conditions — in order. The first one that's True wins. Everything else is ignored.

```python
has_swiggy_one = False
order_value = 250
restaurant_distance_km = 4.5
is_raining = True

if has_swiggy_one:
    delivery_fee = 0
    reason = 'Swiggy One membership'
elif order_value >= 299:
    delivery_fee = 0
    reason = 'Order above ₹299'
elif restaurant_distance_km <= 2:
    delivery_fee = 15
    reason = 'Restaurant nearby'
elif is_raining:
    delivery_fee = 60
    reason = 'Rain surge pricing'
else:
    delivery_fee = 30
    reason = 'Standard fee'

print(f'Delivery Fee : ₹{delivery_fee}')
print(f'Reason       : {reason}')
print(f'Total        : ₹{order_value + delivery_fee}')
```

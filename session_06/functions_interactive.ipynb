## 1️⃣ What is a function?

```python
def verify_aadhaar(aadhaar_number):
    number = int(aadhaar_number)
    if len(str(number)) == 12:
        return 'Valid'
    return 'Invalid'

print(verify_aadhaar('123456789012'))  # Valid
print(verify_aadhaar('1234'))          # Invalid
```

## 2️⃣ Function anatomy — def, name, parameter, return

```python
def check_eligibility(score):   # def + name + parameter
    if score >= 700:            # body
        return True             # return
    return False

print(check_eligibility(742))
print(check_eligibility(600))
```

## 3️⃣ Calling vs defining

```python
def calculate_tip(bill):
    return bill * 0.10

print(calculate_tip(2400))
print(calculate_tip(850))
print(calculate_tip(1200))
```

## 4️⃣ Parameters and arguments

```python
def greet_customer(name, score):
    return f'Hi {name}, your score is {score}'

print(greet_customer('Sambhav', 742))
print(greet_customer('Priya', 810))
```

## 5️⃣ Return keyword

```python
def loan_decision(score, income):
    eligible = score >= 700 and income >= 500000
    return eligible

print(loan_decision(742, 1200000))
print(loan_decision(650, 300000))
```

## 6️⃣ Everything together — Sessions 01–05 inside one function

```python
def process_application(name, score_text, income_text):
    score  = int(score_text)
    income = int(income_text)
    if score >= 700 and income >= 500000:
        return f'{name}: Approved'
    return f'{name}: Rejected'

print(process_application('Sambhav', '742', '1200000'))
print(process_application('Rahul',   '620', '300000'))
```

## 🎯 Try It Yourself

```python
# ✏️ EDIT THESE AND RUN
applicants = [
    ('Sambhav', '742', '1200000'),
    ('Priya',   '810', '800000'),
    ('Rahul',   '620', '300000'),
    ('Ananya',  '700', '500000'),
]

def process_application(name, score_text, income_text):
    score  = int(score_text)
    income = int(income_text)
    if score >= 700 and income >= 500000:
        decision = 'Approved'
    else:
        decision = 'Rejected'
    return f'{name}: {decision} (score={score})'

# ⛔ Don't edit below
print('LOAN APPLICATION RESULTS')
print('=' * 40)
for name, score, income in applicants:
    print(process_application(name, score, income))
```

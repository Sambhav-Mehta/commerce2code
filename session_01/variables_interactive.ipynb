## Variables, Data Types & f-strings

Every time a customer applies for a loan, the system remembers their name, credit score, income.

Let's see how that works.

## 1️⃣ Creating Variables

A variable is just a box where the computer stores information.

```python
# Storing customer data as variables
customer_name = "Sambhav"
credit_score = 742
annual_income = 1200000.00
is_eligible = True

print("Customer data stored in memory:")
print(f"Name: {customer_name}")
print(f"Credit Score: {credit_score}")
print(f"Annual Income: ₹{annual_income:,.2f}")
print(f"Pre-approved: {is_eligible}")
```

## 2️⃣ Understanding Data Types

Each variable has a **type** — text, number, decimal, or yes/no.

```python
# Check the type of each variable
print("Data Types:")
print(f"customer_name is a {type(customer_name).__name__} → {customer_name}")
print(f"credit_score is a {type(credit_score).__name__} → {credit_score}")
print(f"annual_income is a {type(annual_income).__name__} → {annual_income}")
print(f"is_eligible is a {type(is_eligible).__name__} → {is_eligible}")
```

## 3️⃣ f-strings: Displaying Variables in Sentences

An f-string lets you insert variables directly into text. This is how SMS notifications get your name in them.

```python
# Using f-strings to create a real SMS message
if is_eligible:
    sms = f"Congrats {customer_name}! Your score of {credit_score} qualifies you."
else:
    sms = f"Sorry {customer_name}, your score is too low."

print("📱 SMS Notification:")
print(sms)
```

## Operators — Arithmetic, Comparison, Logical

> You learned + - * / in school. Splitwise uses them to split bills. Zomato uses them to calculate delivery fees. CRED uses them to check your cashback eligibility.

## 1️⃣ Arithmetic Operators — the maths you already know
Addition, subtraction, multiplication, division. Exactly like school. Just used at scale.

```python
# How Splitwise splits a dinner bill
bill = 2400
friends = 4

per_person = bill / friends    # standard division
tip        = bill * 0.10       # multiplication
remainder  = bill % friends    # modulo — what's left over after dividing
total      = bill + tip        # addition

print(f'Per person    : ₹{per_person}')
print(f'Tip (10%)     : ₹{tip}')
print(f'Remainder     : ₹{remainder}')
print(f'Total with tip: ₹{total}')
```

## 2️⃣ Comparison Operators — yes/no questions
These always return `True` or `False`. The app is asking a question about your data.

```python
# How CRED checks if you qualify for cashback
credit_score = 742

print(f'credit_score >= 700  →  {credit_score >= 700}')   # greater than or equal
print(f'credit_score == 800  →  {credit_score == 800}')   # exactly equal
print(f'credit_score != 600  →  {credit_score != 600}')   # not equal
print(f'credit_score < 500   →  {credit_score < 500}')    # less than

# Used in an actual decision
if credit_score >= 700:
    print('\n✅ Eligible for cashback')
else:
    print('\n❌ Score too low')
```

## 3️⃣ Logical Operators — combining multiple conditions
Banks don't check just one thing. They check everything at once. Logical operators make that possible.

```python
# How a bank approves a loan
score = 742
income = 1200000
has_existing_loan = False

# and — ALL conditions must be True for approval
approved = score >= 700 and income >= 500000 and not has_existing_loan

print(f'Score OK    : {score >= 700}')
print(f'Income OK   : {income >= 500000}')
print(f'No Loan     : {not has_existing_loan}')
print(f'\nApproved    : {approved}')
```

## 🎯 Try It Yourself
Edit the values below and run. Try failing one condition — watch the approval flip to False.

```python
# ✏️ EDIT THESE VALUES AND RUN
bill               = 2400     # Total bill amount
friends            = 4        # Number of people splitting
credit_score       = 742      # Try 650 — watch eligibility change
income             = 1200000  # Try 400000 — watch approval change
has_existing_loan  = False    # Try True — watch approval flip
has_promo          = True     # Try False — watch discount disappear

# ⛔ Don't edit below this line
print('=' * 48)
print('SPLITWISE')
print('=' * 48)
per_person = bill / friends
print(f'Each person pays: ₹{per_person}')

print('\n' + '=' * 48)
print('CRED CASHBACK')
print('=' * 48)
eligible = credit_score >= 700
print(f'Cashback eligible: {eligible}')

print('\n' + '=' * 48)
print('LOAN APPROVAL')
print('=' * 48)
approved = credit_score >= 700 and income >= 500000 and not has_existing_loan
print(f'Loan approved: {approved}')

print('\n' + '=' * 48)
print('PROMO DISCOUNT')
print('=' * 48)
if per_person > 500 and has_promo:
    discount = per_person * 0.10
    print(f'Discount applied: ₹{discount}')
    print(f'You pay: ₹{per_person - discount:.1f} instead of ₹{per_person}')
else:
    print(f'No discount. Pay: ₹{per_person}')
```

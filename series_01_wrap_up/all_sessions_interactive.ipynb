## Session 01 — Variables & Data Types
Storing information in named boxes.

```python
# ✏️ Edit this list — add your own name and details
applicants = [
    ('Sambhav', '742', '1200000'),
    ('Priya',   '810', '800000'),
    ('Rahul',   '620', '300000'),
    ('Ananya',  '700', '500000'),
]

min_score  = 700
min_income = 500000
bank_name  = 'BFHL Bank'

print(f'Processing {len(applicants)} applications for {bank_name}')
```

## Session 06 — Functions
Write once. Call everywhere.

```python
def process_loan(name, score_text, income_text):
    # Session 05 — Type Conversion
    score  = int(score_text)
    income = int(income_text)
    
    # Session 04 — Operators
    premium  = score >= 750 and income >= 800000
    standard = score >= min_score and income >= min_income
    
    # Session 02 — Control Flow
    if premium:
        status     = 'Approved — Premium'
        loan_limit = income * 5
    elif standard:
        status     = 'Approved — Standard'
        loan_limit = income * 3
    else:
        status     = 'Rejected'
        loan_limit = 0
    
    return {'name': name, 'score': score, 'income': income,
            'status': status, 'loan_limit': loan_limit}

print('Function defined. Ready to process.')
```

## Session 03 — Loops
Process ALL applicants automatically.

```python
approved = 0
rejected = 0

print('RESULTS')
print('=' * 45)

for name, score, income in applicants:
    r = process_loan(name, score, income)
    print(f"{r['name']:10} | Score: {r['score']} | {r['status']}")
    if r['loan_limit'] > 0:
        print(f"{'':10}   Max loan: ₹{r['loan_limit']:,}")
    if r['status'] != 'Rejected':
        approved += 1
    else:
        rejected += 1

print('=' * 45)
print(f'Approved: {approved} | Rejected: {rejected}')
print(f'Approval rate: {(approved/len(applicants))*100:.1f}%')
```

## 🎯 Try It Yourself
Edit and run — make yourself an applicant.

```python
# ✏️ EDIT THESE AND RUN
your_name   = 'Your Name'
your_score  = '720'       # Try 650 — watch it reject
your_income = '600000'    # Try 400000 — watch it reject

# ⛔ Don't edit below
result = process_loan(your_name, your_score, your_income)
print('YOUR APPLICATION RESULT')
print('=' * 40)
print(f"Name   : {result['name']}")
print(f"Score  : {result['score']}")
print(f"Income : ₹{result['income']:,}")
print(f"Result : {result['status']}")
if result['loan_limit'] > 0:
    print(f"Max    : ₹{result['loan_limit']:,}")
```

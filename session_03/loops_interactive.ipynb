## Loops — for, while, break, continue

> Every time you order on Zomato, the app silently pings nearby delivery partners one by one until someone accepts. A loop does that.

## 1️⃣ The for loop — go through a list, do something for each item

```python
# Zomato pinging all nearby delivery partners
delivery_partners = ['Ravi', 'Suresh', 'Amit', 'Priya', 'Kiran']

print('Pinging nearby delivery partners...')
for partner in delivery_partners:
    print(f'📍 Pinging {partner}...')

print('\nAll partners notified!')
```

## 2️⃣ break and continue — controlling the loop

- **continue** — skip this item, move to the next one
- **break** — stop the loop entirely


```python
# Skip unavailable partners, stop at first available
partners = [
    {'name': 'Ravi',   'available': False},
    {'name': 'Suresh', 'available': False},
    {'name': 'Amit',   'available': True},
    {'name': 'Priya',  'available': True},
    {'name': 'Kiran',  'available': True},
]

print('Finding first available partner...')
for partner in partners:
    if not partner['available']:
        print(f"⏭️  {partner['name']} is busy — skipping")
        continue
    print(f"✅ {partner['name']} accepted the order!")
    break
```

## 3️⃣ The while loop — keep going until a condition is met

```python
# OTP verification — 3 attempts then lock
correct_otp = '4829'
max_attempts = 3
attempts = 0
logged_in = False

# Edit these to simulate different scenarios
simulated_inputs = ['1234', '5678', '4829']

print(f'OTP sent. You have {max_attempts} attempts.\n')

while attempts < max_attempts:
    entered_otp = simulated_inputs[attempts]
    attempts += 1
    print(f'Attempt {attempts}: Entered {entered_otp}')

    if entered_otp == correct_otp:
        print('✅ OTP verified! Login successful.')
        logged_in = True
        break
    else:
        remaining = max_attempts - attempts
        if remaining > 0:
            print(f'❌ Wrong OTP. {remaining} attempt(s) left.\n')

if not logged_in:
    print('\n🔒 Account locked. Too many failed attempts.')
```
